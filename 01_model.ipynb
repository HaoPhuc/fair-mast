```python
import numpy as np
import matplotlib.pyplot as plt
import copy
import math
```

```python
def compute_cost_linear_regression(X, y, w, b):
    """
    Compute the cost for linear regression.

    Parameters:
    X: The inputs as arrays with shape (m, n)
    y: The target values with shape (m,)
    w: The weights for the features with shape (n,)
    b: The bias term.

    Returns:
    cost: The cost for linear regression
    """
    m = X.shape[0] # Number of training examples

    # Compute the predictions:
    f_wb = np.dot(X, w) + b

    # Compute the cost:
    cost = (1/(2*m)) * np.sum((f_wb - y)**2)

    return cost
```

```python
def compute_gradient_linear_regression(X, y, w, b):
    """
    Compute the gradient for linear regression.

    Parameters:
    X: The inputs as arrays with shape (m, n)
    y: The target values with shape (m,)
    w: The weights for the features with shape (n,)
    b: The bias term.
    """
    m, n = X.shape #Number of rows/training examples and number of features

    #Compute the predictions:
    f_wb = np.dot(X, w) + b

    #Compute the gradients:
    dj_dw = np.zeros(n)
    dj_db = 0

    for i in range(m):
        err = f_wb[i] - y[i]
        for j in range(n):
            dj_dw[j] += err * X[i, j]
        dj_db += err

    dj_dw = (1/m) * dj_dw
    dj_db = (1/m) * dj_db

    return dj_dw, dj_db
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def compute_cost_logistic_regression(X, y, w, b):
    """
    Compute the cost for logistic regression.

    Parameters:
    X: The inputs as arrays with shape (m, n)
    y: The target values with shape (m,)
    w: The weights for the features with shape (n,)
    b: The bias term

    Returns:
    cost: The cost for logistic regression
    """
    m = X.shape[0] #Number of training examples
    z = np.dot(X, w) + b #Linear combination of inputs and weights
    f_wb = sigmoid(z) #Apply the sigmoid function to get predictions
    
    J = 0
    for i in range(m):
        J += y[i] * np.log(1 + np.exp(-f_wb[i])) + (1 - y[i]) * np.log(1 + np.exp(f_wb[i]))
    cost = -(1/m) * J
    
    return cost
```

```python
def compute_gradient_logistic_regression(X, y, w, b):
    """
    Compute the gradient for logistic regression.

    Parameters:
    X: The inputs as arrays with shape (m, n)
    y: The target values with shape (m,)
    w: The weights for the features with shape (n,)
    b: The bias term

    Returns:
    dj_dw: The gradient of the cost with respect to the weights
    dj_db: The gradient of the cost with respect to the bias
    """
    m, n = X.shape #Number of rows/training examples and number of features
    z = np.dot(X, w) + b #Linear combination of inputs and weights
    f_wb = sigmoid(z) #Apply the sigmoid function to get predictions
    dj_dw = np.zeros(n)
    dj_db = 0
    for i in range(m):
        err = f_wb[i] - y[i]
        for j in range(n):
            dj_dw[j] += err * X[i, j]
    dj_db += err
    dj_dw = (1/m) * dj_dw
    dj_db = (1/m) * dj_db
    return dj_dw, dj_db
```

```python
def gradient_descent(X, y, w_in, b_in, alpha, num_iters):
    """
    Performs batch gradient descent to learn w and b.

    Parameters:
    X: The inputs as arrays with shape (m, n)
    y: The target values with shape (m,)
    w_in: Initial weights for the features with shape (n,)
    b_in: Initial bias
    alpha: Learning rate
    num_iters: Number of iterations to run gradient descent

    Returns:
    w: Updated weights after running
    b: Updated bias after running
    J_history: A list that contains the cost at each iteration
    """
    #Copy the initial parameters to avoid modifying the original ones
    w = copy.deepcopy(w_in)
    b = copy.deepcopy(b_in)
    
    #For plotting the cost function
    J_history = []
    w_history = []
    for i in range(num_iters):
        #Calculate/update the gradient
        dj_dw, dj_db = compute_gradient_linear_regression(X, y, w, b)

        #Update the parameters 
        w -= alpha * dj_dw
        b -= alpha * dj_db

        #Save cost J at each iteration        
        if i < 100000: # prevent resource exhaustion
            J_history.append(compute_cost_linear_regression(X, y, w, b))

        #Print cost and parameters every at intervals 10 times or as many iterations if < 10
        if i % (math.ceil(num_iters / 10))  == 0:
            w_history.append(w)
            print(f"Iteration {i:4d}: Cost {J_history[-1]:8.2f} and w: {w} b: {b}")

    #Plot the cost function with respect to the number of iterations
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(J_history)
    ax.set_xlabel("Iterations")
    ax.set_ylabel("Cost")
    ax.set_title("Cost function vs. Iterations")
    plt.show()
    return w, b, J_history
```
